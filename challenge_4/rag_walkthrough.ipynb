The archive in `corpus/` is already split into short transcript passages. `rag_pipeline.py` ranks them against a query with BM25 (keyword relevance), after any platform, speaker and date filters. Run the three searches below and look at the top passages: **the highest-ranked passage isn't always the best evidence.**

```python
from orca_backend import search_transcript_segments

def show(**kwargs):
    out = search_transcript_segments(**kwargs)
    for r in out['results']:
        print(f"{r['score']:>6}  {r['episode_id']} @ {r['timestamp']}  {r['speaker']}: {r['text']}")
    if not out['results']:
        print(out['note'])
```

**1. A topic with no filters.** Who is actually speaking in each passage? Which ones are a host or another creator *describing* what a candidate said?

```python
show(query='tariffs')
```

**2. The same topic, filtered to one speaker.** Filters narrow the archive before ranking, so the passages left are that person's own words.

```python
show(query='tariffs', speaker='Dana Reyes')
```

**3. A topic the archive never covers.** No passage clears the relevance bar, so nothing comes back. Your skill should then say there's no verified evidence instead of answering from memory.

```python
show(query='cryptocurrency', speaker='Marcus Whitfield')
```

Takeaway for `skill.md`: pass the filters the question gives you, check who is speaking before you quote, and treat an empty result as an answer.
